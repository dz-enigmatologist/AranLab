# Main Function

```python
def main():
    master_folder = fnf.select_folder()
    if not master_folder:
        print("No folder selected. Exiting.")
        return

    output_folder = os.path.join(master_folder, "Processed_Chip_Data")
    os.makedirs(output_folder, exist_ok=True)

    log_file = os.path.join(output_folder, "analysis_log.txt")
    open(log_file, 'w').close()  # Clear previous log

    chip_data = vcf.find_valid_chips(master_folder)
    valid_chips, invalid_chips = fnf.match_chip_files(chip_data)

    fnf.log_and_print("A valid chip must contain the following files: water.csv, buffer.csv, dna.csv, cyst.csv, cas.xlsx", log_file)
    fnf.log_and_print(f"Found {len(chip_data)} chip groups in all subfolders.", log_file)

    if not valid_chips:
        fnf.log_and_print("No valid chips found in the selected folder.", log_file)
        return

    if invalid_chips:
        fnf.log_and_print("Invalid chips found. Missing files:", log_file)
        for chip, missing in sorted(invalid_chips.items(), key=lambda x: int(x[0])):
            missing_types = [key for key in ["water", "buffer", "dna", "cyst", "cas"] if key not in missing]
            missing_list = ", ".join(missing_types)
            fnf.log_and_print(f"Chip {chip} is missing: {missing_list}", log_file)

    fnf.log_and_print(f"Valid chips found: {sorted(valid_chips.keys(), key=int)}", log_file)
    chip_choice = input("Enter chip number to process or 'all' to process all valid chips: ").strip()

    if chip_choice.lower() == "all":
        for chip, files in sorted(valid_chips.items(), key=lambda x: int(x[0])):
            fnf.log_and_print(f"Processing Chip {chip}...", log_file)
            fp.process_chip(chip, files, output_folder)
    else:
        if chip_choice in valid_chips:
            fnf.log_and_print(f"Processing Chip {chip_choice}...", log_file)
            fp.process_chip(chip_choice, valid_chips[chip_choice], output_folder)
        else:
            fnf.log_and_print(f"Chip {chip_choice} not found among valid chips.", log_file)
```

```python
if __name__ == "__main__":
    main()
```
